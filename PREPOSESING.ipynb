```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Data_Game_Acak.csv')
x = dataset.iloc[:, :-1]
y = dataset.iloc[:, -1]
```

```python
display(x.head())
```

```python
display(y.head())
```

```python
from sklearn.impute import SimpleImputer
x_imputed = x.replace('', np.nan)
imputer = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
x_imputed_array = imputer.fit_transform(x_imputed)
x_imputed_df = pd.DataFrame(x_imputed_array, columns=x.columns)
display(x_imputed_df.head())
```

```python
print(f"Jumlah baris di x_imputed_df: {len(x_imputed_df)}")
print(f"Bentuk (rows, columns) x_imputed_df: {x_imputed_df.shape}")
```

```python
from sklearn.impute import SimpleImputer
imputer_reapply = SimpleImputer(missing_values=np.nan, strategy='most_frequent')

# Apply imputer to columns 0 to 3 (which covers all columns in x_imputed_df as it has 3 columns)
x_reimputed_array = imputer_reapply.fit_transform(x_imputed_df.iloc[:, 0:3])

# Convert the numpy array back to a DataFrame, preserving column names
x_reimputed_df = pd.DataFrame(x_reimputed_array, columns=x_imputed_df.iloc[:, 0:3].columns)

display(x_reimputed_df.head())
```

```python
print(f"Jumlah baris di x_reimputed_df: {len(x_reimputed_df)}")
print(f"Bentuk (rows, columns) x_reimputed_df: {x_reimputed_df.shape}")
```

### One-Hot Encoding Fitur (x)

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0, 2])], remainder='passthrough')
x_encoded = ct.fit_transform(x_reimputed_df)
display(x_encoded)
```

### Dataset Splitting (Train/Test)

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x_encoded, y, test_size=0.2, random_state=1)

print("X_train (first 5 rows):")
display(pd.DataFrame(X_train.toarray()).head()) # Convert sparse matrix to dense for display

print("\nX_test (first 5 rows):")
display(pd.DataFrame(X_test.toarray()).head()) # Convert sparse matrix to dense for display

print("\ny_train (first 5 rows):")
display(y_train.head())

print("\ny_test (first 5 rows):")
display(y_test.head())
```

### Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
X_train_dense = X_train.toarray()
X_test_dense = X_test.toarray()
sc = StandardScaler()
X_train_scaled_part = sc.fit_transform(X_train_dense[:, 3:])
X_test_scaled_part = sc.transform(X_test_dense[:, 3:])
X_train_scaled = np.hstack((X_train_dense[:, :3], X_train_scaled_part))
X_test_scaled = np.hstack((X_test_dense[:, :3], X_test_scaled_part))

print("X_train_scaled (first 5 rows):")
display(pd.DataFrame(X_train_scaled).head())

print("\nX_test_scaled (first 5 rows):")
display(pd.DataFrame(X_test_scaled).head())
```
